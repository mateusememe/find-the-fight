```python
import os

ORIGIN_PATH = "..\\datasets"
datasets_path = os.listdir(ORIGIN_PATH)

peliculas_videos_frames_count = []
surv_videos_frames_count = []
LABELS = ["fight", "noFight"]
csv_row_string = ""
```

```python

for dataset in datasets_path:
  out_file = os.path.join(ORIGIN_PATH, dataset, f"{dataset}_path_label_gdrive.csv")

  with open(out_file, 'w') as f:
    f.write("path, label\n")
    for label in LABELS:
      print(os.path.join(ORIGIN_PATH, dataset, label))
      videos_path = os.path.join(ORIGIN_PATH, dataset, label) 
      video_filenames = os.listdir(videos_path)

      for video_path in video_filenames:
        gdrive_path = f"/content/drive/MyDrive/MATT/datasets/Peliculas/{label}/{video_path}"
        # path = os.path.join(GDRIVE_PATH, label, video_path)

        if dataset == "Peliculas":
          csv_row_string = f"{gdrive_path}, {LABELS.index(label)}\n"

        elif dataset == "fight-detection-surv-dataset":
          csv_row_string = f"{gdrive_path},{LABELS.index(label)}\n"
        
        f.write(csv_row_string)
    f.close()
```
