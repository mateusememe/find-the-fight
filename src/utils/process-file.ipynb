```python
import os
import cv2
```

```python
datasets_path = os.listdir("datasets")
ORIGIN_PATH = "datasets"

peliculas_videos_frames_count = []
surv_videos_frames_count = []

for dataset in datasets_path:
  for _class in ["fight", "noFight"]:
    print(os.path.join(ORIGIN_PATH, dataset, _class))
    videos_path = os.path.join(ORIGIN_PATH, dataset, _class) 
    video_filenames = os.listdir(videos_path)

    for video_path in video_filenames:
      path = os.path.join(videos_path, video_path)
      reader = cv2.VideoCapture(path)
      success, frame = reader.read()
      frames_count = 0

      while success:
        success, frame = reader.read()
        frames_count += 1
      

      if dataset == "Peliculas":
        peliculas_videos_frames_count.append({
          "path": os.path.join(videos_path, video_path),
          "frame_count": frames_count
        })
      elif dataset == "fight-detection-surv-dataset":
        surv_videos_frames_count.append({
          "path": os.path.join(videos_path, video_path),
          "frame_count": frames_count
        })


# videos_frames_count
```

```python
len(peliculas_videos_frames_count)
```

```python
classes = {"fight": 0, "noFight": 1}
start_time = 0
```

```python
def make_processed_file(dataset, frames_count):
  annotation_out_file = os.path.join(ORIGIN_PATH, dataset, f"{dataset}_processed.csv")

  with open(annotation_out_file, 'w') as f:
    for video in frames_count:
      path = video["path"]
      end_frame = video["frame_count"]

      _, _dataset, _class, video_id = path.split("\\")
      class_id = classes[_class]
      video_id, _ = video_id.split(".")

      annotation_string = f"{video_id} {start_time} {end_frame} {class_id}\n"
      f.write(annotation_string)
```

```python
datasets_path
```

```python

surveillence, peliculas = datasets_path

make_processed_file(dataset=surveillence, frames_count=surv_videos_frames_count)  
make_processed_file(dataset=peliculas, frames_count=peliculas_videos_frames_count)  
```
